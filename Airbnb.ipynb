##Steps:
1. extract useful data from 'amenities'; amenities column divide on: 30 primary amenities and 'others column' (after train-test split)
2. extract adress column
3. extract features

###1. feature engineering for 'amenities'

```python
from collections import Counter

def parse_amenities(s):
    if pd.isna(s):
        return []

    items = [x.strip() for x in s.split(',')]
    items = [x for x in items if x and not x.startswith('Unavailable')
              and 'Show more' not in x and not x.startswith('󹀁')]
    return items

df['amenities_list'] = df['amenities'].apply(parse_amenities)
df['amenities_count'] = df['amenities_list'].apply(len)
df = df.drop(columns=['amenities'])

# the top-30 amenities and binary amenity_* columns are created AFTER the train/test split
# (fit top-N only on the training set so that the test set does not influence column selection)
```

###2. extract adress column





```python
df.head()
```

```python
# 1. region from address - perform bucketing/dummy encoding AFTER the train/test split
# (fit only on the training set to prevent the top-N categories from "peeking" at the test set)
parts = df['address'].str.split(',')
df['region'] = parts.apply(lambda p: p[-2].strip() if len(p) >= 2 else None)
df['country'] = df['country'].str.strip()

# 2. drop 'adress'
df = df.drop(columns=['address'])
```

```python
df.columns
```

###3. extract features

```python
import pandas as pd
import re

# creating new features
def extract_bathrooms(row):
    s = row['features']
    if pd.isna(s):
        return row['bathrooms']
    m = re.search(r'(\d+(?:\.\d+)?)\s+(?:shared |private )?bathrooms?\b', s)
    return float(m.group(1)) if m else row['bathrooms']

df['bathrooms'] = df.apply(extract_bathrooms, axis=1)

df['bathroom_shared']  = df['features'].str.contains('shared bathroom', na=False).astype(int)
df['bathroom_private'] = df['features'].str.contains('private bathroom', na=False).astype(int)
df['toilet_only'] = df['features'].str.contains('Toilet with sink', na=False).astype(int)
```

```python
df.columns
```

###4. fix checkin/checkout


```python
import numpy as np
# 1. fill the missing values with new category
df['checkin'] = df['checkin'].str.strip().fillna('Unknown')
df['checkout'] = df['checkout'].str.strip().fillna('Unknown')

# 2. top-n categories and dummies after train/test split (fit only on train)
```

###5. fix rating


```python
# 1. "no rating" flag
df['reviews'] = df['reviews'].str.replace(',', '').astype(int)
df['no_rating'] = (df['rating'] == 'New').astype(int)

# 2. convert into numbers, 'New' becomes NaN
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

# 3. fill in missing values ​​with the median after the train/test split—calculating the median based only on the training set
```

###6. drop: 'features', 'id', 'name', 'host_name', 'host_id', 'safety_rules', 'hourse_rules', 'img_links', 'Unnamed: 0', 'address' + 'amenities', 'amenities_list'

```python
df = df.drop(columns=['features', 'id', 'name', 'host_name', 'host_id', 'safety_rules', 'hourse_rules',
                      'img_links', 'Unnamed: 0'])
```

# Building the models

```python
from sklearn.model_selection import train_test_split

X = df.drop('price', axis = 1)
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## Feature engineering continuation

```python
CAT_COLS = ['region', 'country', 'checkin', 'checkout']
TOP_N = {'region': 30, 'country': 20, 'checkin': 5, 'checkout': 5}

def bucket_top_n(train_col, test_col, n):
    top = train_col.value_counts().nlargest(n).index
    train_b = train_col.where(train_col.isin(top), 'Other')
    test_b = test_col.where(test_col.isin(top), 'Other')
    return train_b, test_b

# rating: the median is calculated based only on the training set to prevent test data leakage
rating_median = X_train['rating'].median()
X_train['rating'] = X_train['rating'].fillna(rating_median)
X_test['rating'] = X_test['rating'].fillna(rating_median)

# fix the top-n categories based on the training set and apply them to the test set
for col in CAT_COLS:
    X_train[col], X_test[col] = bucket_top_n(X_train[col], X_test[col], TOP_N[col])

# amenities column divide on 30 primary amenities and 'others column'
AMENITIES_TOP_N = 30
amenities_counter = Counter(a for lst in X_train['amenities_list'] for a in lst)
top_amenities = [a for a, _ in amenities_counter.most_common(AMENITIES_TOP_N)]

for a in top_amenities:
    col = 'amenity_' + a.lower().replace(' ', '_').replace('-', '_')
    X_train[col] = X_train['amenities_list'].apply(lambda lst: int(a in lst))
    X_test[col] = X_test['amenities_list'].apply(lambda lst: int(a in lst))

X_train = X_train.drop(columns=['amenities_list'])
X_test = X_test.drop(columns=['amenities_list'])

# one-hot encoding for models without native category support (RandomForest, XGBoost, LightGBM)
X_train_ohe = pd.get_dummies(X_train, columns=CAT_COLS, drop_first=True)
X_test_ohe = pd.get_dummies(X_test, columns=CAT_COLS, drop_first=True)
X_test_ohe = X_test_ohe.reindex(columns=X_train_ohe.columns, fill_value=False)

# native categories for CatBoost only (without fragmenting the signal into dozens of dummy columns)
X_train_cb = X_train.copy()
X_test_cb = X_test.copy()
for col in CAT_COLS:
    X_train_cb[col] = X_train_cb[col].astype('category')
    X_test_cb[col] = X_test_cb[col].astype('category')
```

```python
from catboost import CatBoostRegressor

cbr = CatBoostRegressor(
    n_estimators=1000,
    learning_rate=0.01,
    depth=6,
    l2_leaf_reg=10,
    loss_function="RMSE",
    eval_metric="MAPE",
    random_strength=1,
    bagging_temperature=1,
    bootstrap_type='Bayesian',
    rsm=0.8,
    random_state=42,
    verbose=200,
    early_stopping_rounds=100
)

cbr.fit(X_train_cb, y_train, eval_set=[(X_test_cb, y_test)], cat_features=CAT_COLS)

cbr.score(X_test_cb, y_test), cbr.score(X_train_cb, y_train)
```

## CatBoost with HalvingRandomSearchCV

```python
from catboost import CatBoostRegressor
from sklearn.experimental import enable_halving_search_cv
from sklearn.model_selection import HalvingRandomSearchCV


grid = {'n_estimators': [500, 1000, 2000],
        'max_depth': [3, 6, 8],
        'l2_leaf_reg': [3, 5, 10],
        'learning_rate': [0.2, 0.01, 0.005]}

cbr_hs = CatBoostRegressor(
    random_state=42,
    cat_features=tuple(CAT_COLS)
)

hs = HalvingRandomSearchCV(estimator=cbr_hs,
                           param_distributions=grid,
                           cv=5,
                           n_jobs=-1,
                           random_state=42)

hs.fit(X_train_cb, y_train)

hs.score(X_test_cb, y_test), hs.score(X_train_cb, y_train)
```

# Evaluation

```python
from sklearn.model_selection import validation_curve

param_range = [1, 3, 5, 10, 20, 30]

train_scores, test_scores = validation_curve(
    estimator=cbr_hs,
    X=X_train_cb,
    y=y_train,
    param_name='l2_leaf_reg',
    param_range=param_range,
    cv=5,
    scoring='r2'
)

train_mean = np.mean(train_scores, axis=1)
train_std = np.std(train_scores, axis=1)
test_mean = np.mean(test_scores, axis=1)
test_std = np.std(test_scores, axis=1)

plt.plot(param_range, train_mean, color='blue', marker='o', markersize=5, label='Training R²')
plt.fill_between(param_range, train_mean + train_std, train_mean - train_std, alpha=0.15, color='blue')

plt.plot(param_range, test_mean, color='green', linestyle='--', marker='s', markersize=5, label='Validation R²')
plt.fill_between(param_range, test_mean + test_std, test_mean - test_std, alpha=0.15, color='green')

plt.grid()
plt.xscale('log')
plt.legend(loc='lower right')
plt.xlabel('l2_leaf_reg')
plt.ylabel('R² score')
plt.show()
```

```python
from sklearn.metrics import mean_absolute_percentage_error, mean_squared_error, r2_score

def evaluate(model, X_train, y_train, X_test, y_test):
  y_preds = model.predict(X_test)
  y_preds_train = model.predict(X_train)

  y_train_real = np.expm1(y_train)
  y_test_real = np.expm1(y_test)
  y_preds_real = np.expm1(y_preds)
  y_preds_train_real = np.expm1(y_preds_train)

  mape = mean_absolute_percentage_error(y_test_real, y_preds_real)
  mse = mean_squared_error(y_test_real, y_preds_real)
  r2_test = r2_score(y_test_real, y_preds_real)
  r2_train = r2_score(y_train_real, y_preds_train_real)
  overfitting = r2_train - r2_test

  return {
      'mape': mape,
      'mse': mse,
      'overfitting': overfitting,
      'r2_test': r2_test
      }

evaluate(model=hs, X_train=X_train_cb, y_train=y_train, X_test=X_test_cb, y_test=y_test)
```
